```python
import torch

a = torch.arange(6)
# aは、0-5までの数字が入ったtensor
print(a)
# a.shapeは、tensorの形(サイズ)を表す。1次元で要素が6個ある
print(a.shape)
```

```python
# reshapeは、Tensorの形(数字の列や行)を変える
b = a.reshape(2,3)

print(b)
print(b.shape)
```
