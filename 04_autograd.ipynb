# 04 Autograd

## 今日も目標

- Autogradとは何かを理解する
- requires_gradの役割を知る
- backward()で勾配を計算してみる

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

print(x)
```

```python
y = x ** 2 # y = xの2乗(2.0^2 = 4.0)

print(y)
```

```python
y.backward() # 微分を実行！ (y を x で微分すると 2x)

print(x.grad)
```

```python
x = torch.tensor(5.0, requires_grad=True)

y = x ** 2 # y = 5.0^2 = 25.0

y.backward() # 微分を実行！

print(x.grad)
```

```python
x = torch.tensor(3.0, requires_grad=True)

y = 3 * x + 2 # y = 3 * 3.0 + 2 = 11.0

y.backward()

print(x.grad)
```

# 今日のまとめ

## 理解できたこと
- requires_gradと指定したTensorは、計算の履歴が自動で記録される
- y.backward()を呼ぶと、yから遡ってxの傾き(変化の割合)を自動で計算してくれる。
- 計算された傾き(勾配)はx.gradの中に保存される
- 傾きはいわゆるslopeのことだが、AIや多次元ではGradient(グラディエント/勾配)と呼ぶ。省略形がgrad。
- ** は「累乗（〜乗）」を表す
- requires_grad=False
追跡しない	
計算結果だけを保持する（高速・省メモリ）。あとで backward() を呼んでも傾きは計算されない。
- grad_fn=<PowBackward0>
PyTorchが裏でメモした「この数字はどうやって計算されたか？」という計算履歴（レシートのようなもの）

fn ➔ Function（関数 / 計算方法）
Pow ➔ Power（累乗 / ** のこと）
Add ➔ Add（足し算）
Mul ➔ Mul（掛け算）
Backward ➔ backward()（微分）するときに使う計算用

## requires_gradとは？
「この数字の変化（勾配）を追跡・記録してください！」というPyTorchへのサイン（設定）。

これを True にしておかないと、後で微分（勾配計算）ができない。
## backward()とは？
「記録された計算を逆再生（バックワード）して、微分（勾配）を計算して！」という実行命令。最後の結果（y）からスタートして、元の数字（x）に向かって遡りながら傾きを計算する。

## gradには何が入っていた？
微分した結果（＝勾配 / 傾き） が入っていた。意味としては 「xをほんの少し増やしたとき、yがどれくらい増えるか（変化の割合）」 。y = x^2 の微分は 2x なので、x=2 のときの x.grad は 4.0
y = 3x + 2 の微分は 3 なので、x=3 のときの x.grad は 3.0
